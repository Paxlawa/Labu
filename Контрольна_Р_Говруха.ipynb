```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
uploaded = files.upload()
df = pd.read_csv('shopping_trends.csv')
df.head()
```

```python

print("Розмір:", df.shape)
df.info()
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.columns
```

```python
plt.figure()
sns.histplot(df['age'], kde=True)
plt.title("Розподіл віку")
plt.show()
```

```python
plt.figure()
sns.countplot(data=df, x='item_purchased', hue='gender')
plt.xticks(rotation=90)
plt.title("Покупки за статтю")
plt.show()
```

```python
plt.figure()
df['item_purchased'].value_counts().head(10).plot(kind='bar')
plt.title("Топ товарів")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='age',
    y='purchase_amount_(usd)',
    hue='gender',
    s=80
)

plt.title('Залежність між віком та сумою покупок')
plt.xlabel('Вік')
plt.ylabel('Сума покупок (USD)')
plt.grid(True)
plt.show()
```

```python
plt.figure()
sns.histplot(df['review_rating'], kde=True)
plt.title("Рейтинг товарів")
plt.show()
```

```python
plt.figure()
sns.countplot(data=df, x='gender')
plt.title("Розподіл за статтю")
plt.show()
```

```python
plt.figure()
sns.countplot(data=df, x='gender', hue='item_purchased')
plt.xticks(rotation=90)
plt.title("Покупки за статтю")
plt.show()
```

```python
plt.figure()
sns.scatterplot(data=df, x='review_rating', y='purchase_amount_(usd)')
plt.title("Рейтинг vs Сума покупки")
plt.show()
```

```python
import altair as alt
from vega_datasets import data

wheat = data.wheat()

chart1 = alt.Chart(wheat).mark_bar().encode(
    x=alt.X('wheat:Q', title='Врожайність пшениці'),
    y=alt.Y('year:O', title='Рік', sort='-x'),
    tooltip=['year', 'wheat']
).properties(
    title='Врожайність пшениці по роках'
)

chart1
```

```python
iowa = data.iowa_electricity()

chart2 = alt.Chart(iowa).mark_bar(opacity=0.7).encode(
    x=alt.X('year:T', title='Рік'),
    y=alt.Y('net_generation:Q', title='Чисте виробництво'),
    color=alt.Color('source:N', title='Джерело енергії'),
    tooltip=['year', 'net_generation', 'source']
).properties(
    title='Виробництво електроенергії в Айові'
)

chart2
```

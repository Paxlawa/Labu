```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 200)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, y1, linestyle='-', marker='o', markevery=20, label='sin(x)')
ax.plot(x, y2, linestyle='--', marker='s', markevery=20, label='cos(x)')

ax.set_title("Дві лінії")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```

Побудуйте contour для функції Z(X,Y) на сітці

```python

x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)
plt.figure(figsize=(6, 5))
plt.contour(X, Y, Z, levels=20, cmap='viridis')
plt.title("Contour plot Z(X,Y)")
plt.xlabel("X")
plt.ylabel("Y")
plt.colorbar(label="Z")
plt.grid(alpha=0.3)
plt.show()
```

Побудуйте 3D-поверхню Z = sin(sqrt(X²+Y²))

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Сітка
x = np.linspace(-6, 6, 200)
y = np.linspace(-6, 6, 200)
X, Y = np.meshgrid(x, y)

# Функція
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='plasma', edgecolor='none')
ax.set_title("3D поверхня Z = sin(sqrt(X² + Y²))")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

Dashboard з GridSpec (4 графіки + заголовок)

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

np.random.seed(0)


t = np.linspace(0, 20, 300)
trend = 0.15*t + np.sin(t) + np.random.normal(0, 0.3, len(t))

x = np.random.randn(400)
y = 0.8*x + np.random.randn(400)/2

indicator_values = [30, 55, 80, 45]
indicator_labels = ["A", "B", "C", "D"]


fig = plt.figure(figsize=(12, 7))
gs = GridSpec(3, 3, figure=fig, width_ratios=[2, 2, 1])


fig.suptitle("Dashboard з GridSpec", fontsize=16)


ax1 = fig.add_subplot(gs[0, :2])
ax1.plot(t, trend, linewidth=2)
ax1.set_title("Тренд (line)")
ax1.set_xlabel("t")
ax1.set_ylabel("trend")
ax1.grid(alpha=0.3)


ax2 = fig.add_subplot(gs[1:, 0])
ax2.hist(x, bins=25, color='orange', alpha=0.8)
ax2.set_title("Гістограма")
ax2.set_xlabel("x")
ax2.set_ylabel("count")
ax2.grid(alpha=0.3)


ax3 = fig.add_subplot(gs[1:, 1])
ax3.scatter(x, y, alpha=0.6, color='red')
ax3.set_title("Scatter")
ax3.set_xlabel("x")
ax3.set_ylabel("y")
ax3.grid(alpha=0.3)


ax4 = fig.add_subplot(gs[:, 2])
ax4.barh(indicator_labels, indicator_values, color=['green', 'blue', 'purple', 'gray'])
ax4.set_title("Індикатор")
ax4.set_xlabel("value")
ax4.grid(alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```

анімація руху точки по sin(x)

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

x = np.linspace(0, 2*np.pi, 400)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y, color='blue')

point, = ax.plot([], [], 'ro', markersize=8)
text = ax.text(0.05, 0.9, "", transform=ax.transAxes)

ax.set_xlim(0, 2*np.pi)
ax.set_ylim(-1.2, 1.2)
ax.grid(True)

def init():
    point.set_data([], [])
    text.set_text("")
    return point, text

def update(i):
    px = x[i]
    py = y[i]
    point.set_data(px, py)
    text.set_text(f"x={px:.2f}, y={py:.2f}")
    return point, text

ani = FuncAnimation(fig, update, frames=len(x), init_func=init,
                    interval=20, blit=True)

plt.show()
```

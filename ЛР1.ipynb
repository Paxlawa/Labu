```python
a = np.array([1, 2])
b = np.array([2, 1])

cos_angle = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

angle_rad = np.arccos(cos_angle)
angle_deg = np.degrees(angle_rad)

print("Завдання 25:")
print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle_deg)
print("-" * 40)
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[2, 0],
              [1, 2]])

elem_mult = A * B
matmul = A @ B

print("Завдання 27:")
print("Поелементний добуток (A * B):\n", elem_mult)
print("Матричний добуток (A @ B):\n", matmul)

print("-" * 40)
```

```python
A = np.array([[4, 7],
              [2, 6]])

det = np.linalg.det(A)
A_inv = np.linalg.inv(A)


check = A @ A_inv

print("Завдання 28:")
print("Визначник:", det)
print("Обернена матриця:\n", A_inv)
print("Перевірка (A @ A_inv):\n", check)
print("-" * 40)
```

```python
np.random.seed(42)
grades = np.random.randint(0, 101, size=30)

avg_grade = np.mean(grades)
median_grade = np.median(grades)
max_grade = np.max(grades)
min_grade = np.min(grades)

count_60_plus = np.sum(grades >= 60)
pct_90_plus = (np.sum(grades >= 90) / len(grades)) * 100

print("Завдання 35:")
print("Оцінки:", grades)
print("Середня оцінка:", avg_grade)
print("Медіана:", median_grade)
print("Найвища оцінка:", max_grade)
print("Найнижча оцінка:", min_grade)
print("Кількість студентів (>= 60):", count_60_plus)
print("Відсоток студентів (>= 90):", pct_90_plus, "%")
print("-" * 40)
```

```python
sales = np.array([[120, 150, 170, 160],
                  [80,  90,  100, 110],
                  [200, 180, 210, 230]])

total_per_product = np.sum(sales, axis=1)
total_per_month = np.sum(sales, axis=0)

best_product_idx = np.argmax(total_per_product) + 1
best_month_idx = np.argmax(total_per_month) + 1

avg_per_product = np.mean(sales, axis=1)

print("Завдання 36:")
print("Загальний обсяг продажів кожного товару:", total_per_product)
print("Загальний обсяг продажів у кожному місяці:", total_per_month)
print(f"Товар із найбільшими продажами: Товар №{best_product_idx}")
print(f"Місяць із найбільшим обсягом продажів: Місяць №{best_month_idx}")
print("Середній обсяг продажів кожного товару:", avg_per_product)
print("-" * 40)
```

```python
np.random.seed(42)
M = np.random.randint(1, 21, size=(5, 5))

main_diag_sum = np.trace(M)
side_diag_sum = np.trace(np.fliplr(M))

max_per_row = np.max(M, axis=1)
min_per_col = np.min(M, axis=0)

M_modified = M.copy()
M_modified[M_modified < np.mean(M)] = 0

print("Завдання 37:")
print("Початкова матриця:\n", M)
print("Сума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", side_diag_sum)
print("Максимуми в кожному рядку:", max_per_row)
print("Мінімуми в кожному стовпці:", min_per_col)
print("Матриця після заміни елементів (< mean) на 0:\n", M_modified)
print("-" * 40)
```

```python
X = np.array([[2, 4, 6],
              [1, 3, 5],
              [5, 7, 9]])

dist_1_2 = np.linalg.norm(X[0] - X[1])
dist_1_3 = np.linalg.norm(X[0] - X[2])
dist_2_3 = np.linalg.norm(X[1] - X[2])

print("Завдання 38:")
print("Відстань між 1 та 2 рядками:", dist_1_2)
print("Відстань між 1 та 3 рядками:", dist_1_3)
print("Відстань між 2 та 3 рядками:", dist_2_3)
print("-" * 40)
```

```python
np.random.seed(42)
A = np.random.randint(1, 10, size=(4, 4))

row_norms = np.linalg.norm(A, axis=1, keepdims=True)

A_normalized = A / row_norms

print("Завдання 39:")
print("Початкова матриця:\n", A)
print("Нормалізована матриця (кожен рядок одиничної довжини):\n", A_normalized)
print("-" * 40)
```

```python
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
b_check = A @ x

print("Завдання 40:")
print("Знайдений розв'язок (x, y, z):", x)
print("Результат A @ x:", b_check)
print("Чи збігається A @ x з b?:", np.allclose(b_check, b))
```
